```python
import cv2
import numpy as np
from ultralytics import YOLO
```

```python
model = 'yolov8n.pt'
model = YOLO(model)

```

```python
capture = cv2.VideoCapture(0)

while True:
    active, frame = capture.read()

    prediction = model(frame) # the detection of the current frame

    cv2.imshow('Capture', prediction[0].plot())
    if cv2.waitKey(1) == 32: # 1ms for video
        break

cv2.destroyAllWindows()
```
